```python
import matplotlib.pyplot as plt

sns.histplot(df["fare"], kde=True)
plt.title('Fare distribution')
plt.show()
```

```python
sns.histplot(df["distance"], kde=True)
plt.title("Distance Distribution")
plt.show()
```

```python
sns.countplot(data=df, x="passengers")
plt.title("Passenger Count Distribution")
plt.show()
```

```python
df[df['passengers']==0].head(10)
```

```python
sns.scatterplot(data=df, x="distance", y="fare", alpha=0.4)
plt.title("Distance vs Fare")
plt.show()
```

```python
sns.scatterplot(data=df, x="fare", y="tip", alpha=0.4)
plt.title("Fare vs Tip")
plt.show()
```

```python
cols = ['distance','tip','fare','tolls']

corr_matrix = df[cols].corr()

sns.heatmap(corr_matrix, annot=True, cmap='Greens')
plt.title("Correlation Heatmap")
plt.show()
```

```python
df["pickup"] = pd.to_datetime(df["pickup"])
df["hour"] = df["pickup"].dt.hour
df["day"] = df["pickup"].dt.day_name()

sns.countplot(data=df, x="hour")
plt.title("Trips by Hour")
plt.show()
```
