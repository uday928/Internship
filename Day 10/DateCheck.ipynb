```python
import pandas as pd
```

```python
dates={
    'name':['a','b','c','d','e','f','g','h','i','j'],
    'date':['17-01-2004','abcd','03/05/12','12-13-2004','01-08-05','2004-11-31','','2006-02-29','15/07/2003','']
}
```

```python
df=pd.DataFrame(dates)
```

```python
df
```

```python
df['date']=pd.to_datetime(df['date'],format='mixed',errors='coerce')
```

```python
df
```

```python
df[df['date'].isna()]
```

```python
df.dropna(subset='date',inplace=True)
```

```python
df=df.reset_index()
```

```python
df.drop(columns=['index'],inplace=True)
```
