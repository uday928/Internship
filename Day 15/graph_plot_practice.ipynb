```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Task 1: plot a line graph of days->temperature and show the maximum and minimum temperature
x=np.arange(1,31)
```

```python
x
```

```python
from numpy import random
y=random.randint(20,40,size=30)
y
```

```python
max_indexs=np.argmax(y)
max_indexs
```

```python
# Max indexes
max_x = x[max_indexs]
max_y = y[max_indexs]
print(max_x,max_y)
```

```python
min_indexs=np.argmin(y)
min_indexs
```

```python
# Min indexes
min_x=x[min_indexs]
min_y=y[min_indexs]
print(min_x,min_y)
```

```python
# Main graph
plt.plot(x,y,marker='o',mfc='white')
plt.scatter(max_x, max_y, color='red',s=100,zorder=3)
plt.scatter(min_x,min_y,color='green',s=90,zorder=3)
plt.xlabel('Days')
plt.ylabel('Temperature')
plt.title('City temperature of this month')
plt.show()
```

```python
# Task 3: Random runs for T20 match
over=np.arange(1,21)
over
```

```python
runs=random.randint(0,18,size=20)
runs
```

```python
total_runs=sum(runs)
print("Total runs: ",total_runs)
```

```python
plt.plot(over,runs)
plt.xlabel('Overs')
plt.ylabel('Runs')
plt.title('Runs per over')
plt.show()
```

```python
# Team wise runs per over:
t1=random.randint(0,198,size=20)
t1_1=np.sort(t1)
t1_1
```

```python
t2=random.randint(0,201,size=20)
t2_1=np.sort(t2)
t2_1
```

```python
plt.plot(over,t1_1,label='Royal Challangers Bangalore',marker='o',ms=4)
plt.plot(over,t2_1,label='Gujarat Titans',marker='o',ms=4)
plt.xticks(over)
plt.xlabel('overs')
plt.ylabel('runs')
plt.title('Match summary')
plt.legend(loc='upper left')
plt.show()
```
