```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.linspace(0, 5 * np.pi)
y = np.sin(x)
```

```python
fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

```python
x = np.linspace(0, 5 * np.pi)
y = np.cos(x)
```

```python
fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

```python
x = np.linspace(0, 5 * np.pi)
y = np.tan(x)
```

```python
fig, ax = plt.subplots()
ax.plot(x, y)
plt.show()
```

```python
# Straight line graph
x=np.array([0,10])
y=np.array([0,250])

plt.plot(x,y)
plt.show()
```

```python
x=np.array([1,2,6,9,10])
y=np.array([-1,0,35,-15,12])

plt.plot(x,y)
plt.show()
```

```python
# To get data points instead of line
x=np.array([1,2,6,9,10])
y=np.array([-1,0,35,-15,12])

plt.plot(x,y,'o')
plt.show()
```

```python
# Default x points
ypoints=np.array([1,2,6,9,10])
plt.plot(ypoints)
plt.show()
```

```python
ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints)
plt.show()
```

```python
# Change in shape of points:
plt.plot(ypoints, 'o')
plt.show()
```

```python
# Size of markers
plt.plot(ypoints, 'h',ms=20)
plt.show()
```

```python
# Colour
plt.plot(ypoints, 'h-r')
plt.show()
```

```python
plt.plot(ypoints, 'h:m')
plt.show()
```

```python
# Markeredge color to make the edge of data points coloring
plt.plot(ypoints, 'o-w',markeredgecolor='k',ms=12)
plt.show()
```

```python
plt.plot(ypoints, 'o:w',mec='g',ms=12)
plt.show()
```

```python
# linestyle
# plt.plot(ypoints,ls='dotted')
plt.plot(ypoints,ls=':')
plt.show()
```

```python
# Line color
plt.plot(ypoints,ls='-',c='hotpink')
plt.show()
```

```python
# width of line
plt.plot(ypoints,ls='-',linewidth=10,c='hotpink',marker='o',mfc='b',ms=18)
plt.show()
```

```python
plt.plot(x)
plt.plot(y)
```

```python
# Titles and labels
font1 = {'family':'sans','color':'blue','size':13}
font2 = {'family':'sans','color':'darkred','size':10}
plt.title('my_test_plot',fontdict=font1,loc='left')
plt.plot(x)
plt.plot(y)
plt.xlabel('x-axis',fontdict=font2)
plt.ylabel('y-axis',fontdict=font2)
```

```python
# grids
font1 = {'family':'sans','color':'blue','size':13}
font2 = {'family':'sans','color':'darkred','size':10}
plt.title('my_test_plot',fontdict=font1,loc='left')
plt.plot(x)
plt.plot(y)
# plt.grid()
# plt.grid(axis='y')
plt.grid(color = 'green', linestyle = '--', linewidth = 0.5)
plt.xlabel('x-axis',fontdict=font2)
plt.ylabel('y-axis',fontdict=font2)
```
