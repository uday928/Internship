```python
plt.figure(figsize=(40,20))
plt.rcParams.update({'font.size':27})
plt.plot(df['stu_id'],df['percentage'],marker='o',ms=20,mfc='red',linewidth=15)
plt.xticks(df['stu_id'])
plt.xlabel('stu_ids')
plt.ylabel('percentages')
plt.title('Performance graph of students (by Pr)')
plt.savefig('dot1',dpi=500)
plt.show()
```

```python
grade_count=df['grade'].value_counts()
grade_count
```

```python
grades=['A','C','B','D']
```

```python
plt.pie(grade_count,labels=grades, autopct='%1.1f%%')
plt.axis('equal')
plt.legend()
plt.show()
```
